### 3. Basic Chat Completion & Cost Tracking

```python
from nroutersdk import nRouter

with nRouter() as client:
    response = client.chat.completions.create(
        model="anthropic/claude-sonnet-4-5-20250929",
        messages=[
            {"role": "system", "content": "You are a helpful AI assistant."},
            {"role": "user", "content": "Explain why multi-provider AI gateways are valuable in 2 sentences."}
        ],
        max_tokens=200
    )
    
    print("Assistant Response:")
    print(response.choices[0].message.content)
    
    # Inspect automatic metadata
    meta = client.last_response
    print("\n--- Request Metadata ---")
    print(f"Request ID:   {meta.request_id}")
    print(f"Exact Cost:   ${meta.cost:.6f}" if meta.cost is not None else "Cost: unpriced")
    print(f"Tokens:       {meta.input_tokens} in / {meta.output_tokens} out")
```

### 4. Real-time Token Streaming

```python
with nRouter() as client:
    stream = client.chat.completions.create(
        model="anthropic/claude-sonnet-4-5-20250929",
        messages=[{"role": "user", "content": "Write a haiku about intelligent cloud routing."}],
        stream=True,
    )
    
    for chunk in stream:
        if chunk.choices and chunk.choices[0].delta.content:
            print(chunk.choices[0].delta.content, end="", flush=True)
    print()
```

### 5. Native Anthropic Messages API

```python
with nRouter() as client:
    # Pre-call token counting
    count = client.messages.count_tokens(
        model="anthropic/claude-sonnet-4-5-20250929",
        messages=[{"role": "user", "content": "How many tokens is this prompt?"}],
    )
    print(f"Input token count: {count['input_tokens']}")
    
    # Call Messages API
    msg = client.messages.create(
        model="anthropic/claude-sonnet-4-5-20250929",
        messages=[{"role": "user", "content": "Hello via Anthropic Messages!"}],
        max_tokens=150
    )
    print("\nResponse:", msg["content"][0]["text"])
```
